## Package import

```python
import pandas as pd
from scipy import stats
from statsmodels.stats import weightstats as stests
```

## Test for Single Mean

```python
# Import: Rossmann Store Sales Data 

sales_data=pd.read_csv("https://raw.githubusercontent.com/venkatareddykonasani/Datasets/master/Rossmann_Store_Sales/Rossmann_Store_Sales.zip", compression="zip")
print(sales_data.shape)
print(sales_data.info())
sales_data.sample(5)
```

```python
# Testing the Null Hypothesis: Average number of customers = 700

sales_sample=sales_data.sample(n=100,random_state=55)   # n = 100 means Sample size = 100

ztest, pval = stests.ztest(sales_sample["Customers"], x2=None, value=700)   # returns a tuple having ztest and p- value
print(ztest)
print("P-Value", float(pval))

if pval<0.05:
    print("reject null hypothesis")
else:
    print("accept null hypothesis")
    
# Null Hypothersis will be accepted for avg no. of customers = 650
```

```python
# Testing the Null Hypothesis: Average sales = 5600

sales_sample=sales_data.sample(n=100,random_state=55)

ztest ,pval = stests.ztest(sales_sample["Sales"], x2=None, value=5600)
print(ztest)
print("P-Value", float(pval))

if pval<0.05:
    print("reject null hypothesis")
else:
    print("accept null hypothesis")
```

## Test for Two Means

```python
sales_sample=sales_data.sample(n=100,random_state=55)

# ztest(sample1, sample2) performs a two-sample Z-test and check : Is the mean sales of Promo = 1 significantly different from Promo = 0?
# H₀ (Null hypothesis): Mean(Sales | Promo=0) = Mean(Sales | Promo=1)
# H₁ (Alternative hypothesis): Mean(Sales | Promo=0) ≠ Mean(Sales | Promo=1)
ztest ,pval = stests.ztest(sales_sample["Sales"][sales_sample["Promo"]==0], 
                            sales_sample["Sales"][sales_sample["Promo"]==1])
print(ztest)
print("P-Value", float(pval))

if pval<0.05:
    print("reject null hypothesis")
else:
    print("accept null hypothesis")
```

## Testing Multiple Means
Using **F-tests** : which is used to compare mean across 3 or more groups

```python
sales_sample=sales_data.sample(n=300,random_state=55)

# to check whether average sales differ across different store types
# Null hypothesis (H₀): All store types have the same mean sales
# Alternative hypothesis (H₁): At least one store type has a different mean

F, p = stats.f_oneway(sales_sample["Sales"][sales_sample["StoreType"]=="c"],
                        sales_sample["Sales"][sales_sample["StoreType"]=="b"],
                        sales_sample["Sales"][sales_sample["StoreType"]=="d"] 
                    )
print(F)  # Small F → group means are similar
print("p-value for significance is: ", p)

if p<0.05:
    print("reject null hypothesis")
else:
    print("accept null hypothesis")
```

```python
# Null hypothesis (H₀): Average number of customers is the same on all selected days
# Alternative hypothesis (H₁): At least one day has a different average number of customers
F, p = stats.f_oneway(sales_sample["Customers"][sales_sample["DayOfWeek"]==2],
                        sales_sample["Customers"][sales_sample["DayOfWeek"]==3],
                        sales_sample["Customers"][sales_sample["DayOfWeek"]==4])

print(F)
print("p-value for significance is: ", p)

if p<0.05:
    print("reject null hypothesis")
else:
    print("accept null hypothesis")
```

## Chi-Square Test of Independence

```python
bank_data=pd.read_csv("https://raw.githubusercontent.com/venkatareddykonasani/python_5days/master/Day3_Files/Datasets/bank-full.csv",sep=';')
print(bank_data.shape)
bank_data.info()
```

```python
bank_data.sample(10)
```

```python
categorical_cols=[col for col in bank_data.columns if bank_data[col].dtypes in ["str"]]
print(categorical_cols)
```

```python
bank_data_sample=bank_data.sample(300)
import scipy.stats as stats

# checking whether each categorical column matter for predicting the target variable y
# H₀ (Null hypothesis): The categorical feature and target y are independent
# H₁ (Alternative hypothesis): The feature and y are dependent (related) -> feature can be dropped
for col in categorical_cols:
    crosstab =pd.crosstab(bank_data_sample[col], bank_data_sample['y'])
    print(crosstab)
    
    Chi=stats.chi2_contingency(crosstab)   # This returns 4 values: (chi2_stat, p_value, dof, expected_freq)
    p=Chi[1]   # getting p_value
    print("p-value for significance is: ",p )
    if p<0.05:
        print("reject null hypothesis")
    else:
        print("accept null hypothesis")
    print("============================\n")
```
