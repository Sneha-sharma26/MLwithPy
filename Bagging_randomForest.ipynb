```python
import pandas as pd
import sklearn as sk
import numpy as np
import scipy as sp
from sklearn import tree
from sklearn.metrics import confusion_matrix
```

### Data Importing

```python
#Importing dataset
car_train=pd.read_csv("https://raw.githubusercontent.com/venkatareddykonasani/Datasets/master/Car_Accidents_IOT/train.csv")
car_train.info()
car_test=pd.read_csv("https://raw.githubusercontent.com/venkatareddykonasani/Datasets/master/Car_Accidents_IOT/test.csv")
```

```python
features=list(car_train.columns[1:])    # excluding the target var "Fatal"
X_train=car_train[features]
y_train=car_train['Fatal']
```

## Decision Tree

```python
# Buildng Decision tree on the training data
clf = tree.DecisionTreeClassifier(max_depth=8)
clf.fit(X_train, y_train)

# Predicting on train and test data
tree_predict_train=clf.predict(car_train[features])
tree_predict=clf.predict(car_test[features])
```

### Decision Tree Results

```python
# Train Data Validation
cm1 = confusion_matrix(y_train,tree_predict_train)
print("Confusion Matrix for Train data \n", cm1)
total1=sum(sum(cm1))
accuracy_tree=(cm1[0,0]+cm1[1,1])/total1
print("Accuracy of Train data \n", accuracy_tree)

# Test Data Validation
cm1 = confusion_matrix(car_test[['Fatal']],tree_predict)
print("Confusion Matrix for test data \n", cm1)
total1=sum(sum(cm1))
accuracy_tree=(cm1[0,0]+cm1[1,1])/total1
print("Accuracy of test data \n", accuracy_tree)

# having max_depth=6, we are getting accuracy of 86% on train data and 85% on test data
# having max_depth=8, we are getting accuracy of 89% on train data and 88% on test data
```

## Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
forest=RandomForestClassifier(n_estimators=100, max_features=7, max_depth=11)
# n_estimators - no. of decision Trees - As high as possible
# max_features - no of features in building the decision trees model (p) - low ==>(sqrt(t))
# max_depth - Slightly higher than single tree
forest.fit(X_train,y_train)

# Predicting on train and test data 
Forest_predict_train=forest.predict(car_train[features])
Forest_predict=forest.predict(car_test[features])
```

### Random Forest Results

```python
#Train Data Validation
cm1 = confusion_matrix(y_train,Forest_predict_train)
print("Confusion Matrix for Train data \n", cm1)
total1=sum(sum(cm1))
accuracy_tree=(cm1[0,0]+cm1[1,1])/total1
print("Accuracy of Train data \n", accuracy_tree)

########Test Data Validation
cm1 = confusion_matrix(car_test[['Fatal']],Forest_predict)
print("Confusion Matrix for test data \n", cm1)
total1=sum(sum(cm1))
accuracy_tree=(cm1[0,0]+cm1[1,1])/total1
print("Accuracy of test data \n", accuracy_tree)

# having n_estimators = 50, max_features=5, max_depth=12, we are getting accuracy of 85% on train data and 85% on test data
# having n_estimators = 100, max_features=5, max_depth=12, we are getting accuracy of 94% on train data and 91% on test data
# having n_estimators = 100, max_features=7, max_depth=11, we are getting accuracy of 93% on train data and 91% on test data
```
